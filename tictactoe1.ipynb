```python
# ============================================================
# CODSOFT AI INTERNSHIP - TASK 2
# TIC-TAC-TOE AI
# Human (X) vs AI (O)
# Algorithm: Minimax
# ============================================================

import random

# ------------------------------------------------------------
# 1. BOARD
# ------------------------------------------------------------

board = [" " for _ in range(9)]


# ------------------------------------------------------------
# 2. DISPLAY BOARD
# ------------------------------------------------------------

def display_board():
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")


# ------------------------------------------------------------
# 3. CHECK WINNER
# ------------------------------------------------------------

def check_winner(b):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b1, c in winning_combinations:
        if b[a] != " " and b[a] == b[b1] == b[c]:
            return b[a]

    if " " not in b:
        return "Draw"

    return None


# ------------------------------------------------------------
# 4. MINIMAX ALGORITHM
# ------------------------------------------------------------

def minimax(b, maximizing):
    result = check_winner(b)

    # Terminal states
    if result == "O":
        return 1
    elif result == "X":
        return -1
    elif result == "Draw":
        return 0

    # AI's turn
    if maximizing:
        best_score = -float("inf")

        for i in range(9):
            if b[i] == " ":
                b[i] = "O"

                score = minimax(b, False)

                b[i] = " "

                best_score = max(best_score, score)

        return best_score

    # Human's turn
    else:
        best_score = float("inf")

        for i in range(9):
            if b[i] == " ":
                b[i] = "X"

                score = minimax(b, True)

                b[i] = " "

                best_score = min(best_score, score)

        return best_score


# ------------------------------------------------------------
# 5. AI MOVE
# ------------------------------------------------------------

def ai_move():
    best_score = -float("inf")
    best_move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            score = minimax(board, False)

            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = "O"


# ------------------------------------------------------------
# 6. HUMAN MOVE
# ------------------------------------------------------------

def human_move():
    while True:
        try:
            move = int(input("Enter your position (1-9): ")) - 1

            if move < 0 or move > 8:
                print("❌ Please enter a number between 1 and 9.")
                continue

            if board[move] != " ":
                print("❌ That position is already occupied.")
                continue

            board[move] = "X"
            break

        except ValueError:
            print("❌ Please enter a valid number.")


# ------------------------------------------------------------
# 7. GAME
# ------------------------------------------------------------

def play_game():

    global board

    board = [" " for _ in range(9)]

    print("=" * 45)
    print("       🎮 TIC-TAC-TOE AI")
    print("       CODSOFT AI INTERNSHIP")
    print("=" * 45)

    print("\nYou are X")
    print("AI is O")

    print("\nBoard Positions:")
    print("""
     1 | 2 | 3
    ---+---+---
     4 | 5 | 6
    ---+---+---
     7 | 8 | 9
    """)

    while True:

        # Human turn
        display_board()
        human_move()

        result = check_winner(board)

        if result:
            display_board()

            if result == "X":
                print("🎉 Congratulations! You won!")

            elif result == "Draw":
                print("🤝 Game Draw!")

            return

        # AI turn
        print("🤖 AI is thinking...")

        ai_move()

        result = check_winner(board)

        if result:
            display_board()

            if result == "O":
                print("🤖 AI wins! Better luck next time!")

            elif result == "Draw":
                print("🤝 Game Draw!")

            return


# ------------------------------------------------------------
# 8. START GAME
# ------------------------------------------------------------

play_game()
```
